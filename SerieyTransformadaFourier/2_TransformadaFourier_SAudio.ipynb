# Transformada de Fourier

 - Se presentan los conceptos básicos de la transformada de Fourier, en sus versiones continua, tiempo discreto y transformada discreta (extendida a su versión rápida).

- La transformada de Fourier se puede entender como la extensión de la serie de Fourier para $t\in(-\infty,+\infty)$:

\begin{equation}
X(\omega) = \mathscr{F} \{x(t) \} = \int_{-\infty}^{+\infty}{x(t)e^{-j\omega t}dt}.
\end{equation}

- La respectiva transformada inversa (extensión de la reconstrucción mediante serie de Fourier con $t\in(-\infty,+\infty)$) se define como:

\begin{equation}
x(t) = \mathscr{F}^{-1} \{X(\omega) \} = \frac{1}{2\pi}\int_{-\infty}^{+\infty}{X(\omega)e^{j\omega t}d\omega}.
\end{equation}

- Cabe resaltar entonces que el concepto de espectro en frecuencia para la transformada de Fourier asume que $\omega \in (-\infty,+\infty).$

- En resumen, la serie de Fourier analiza $x(t)\in \mathbb{R},\mathbb{C}$ en tiempo continuo sobre segmento finito $t\in [-T/2,T/2]$ o $t\in [t_i,t_f]$, en espectro en frecuencia discreto $c_n \in \mathbb{R},\mathbb{C}$, con $n\in \mathbb{Z}.$

- La transformada de Fourier analiza $x(t)\in \mathbb{R},\mathbb{C}$ en tiempo continuo sobre segmento infinito de tiempo $t\in (-\infty,+\infty)$, en espectro en frecuencia continuo (función de densidad espectral) $X(\omega)\in \mathbb{R},\mathbb{C}$, con $\omega\in (-\infty,+\infty).$


- Las propiedades principales de la transformada de Fourier son útiles para analizar el espectro en fenómenos físicos clásicos (Ver material apoyo [propiedades de la transformada de Fourier](https://github.com/amalvarezme/SenalesSistemas/blob/master/3_SerieyTransformadaFourier/Propiedades_TransformadaFourier.pdf)).

- Además, algunas de las transformadas básicas se pueden encontrar en las [Tablas de la transformada de Fourier](https://github.com/amalvarezme/SenalesSistemas/blob/master/3_SerieyTransformadaFourier/Tablas_TransformadaFourier.pdf).

# Transformada de Fourier en tiempo discreto

- A partir de la transformada de Fourier, se puede analizar también una señal en tiempo discreto $x[k]\in \mathbb{R} ,\mathbb{C},$ con $k\in\mathbb{Z}$, con el fin de calcular su espectro en frecuencia continuo (función de densidad espectral) $X(\omega)\in \mathbb{R} ,\mathbb{C}$, pero con $\omega$ continua en el intervalo $\omega\in [-\pi,+\pi]$ o $\omega\in [0,2\pi],$ dado que las bases de representación requieren restringir la codificación de la señal $x[k]$ utilizando funciones seno y coseno originales según la expresión:

\begin{equation}
X(\omega) = \sum_{k\in\mathbb{Z}}{x[k]e^{-j\omega k}},
\end{equation}

- **Nota**: Recuerde que cos/sen en tiempo discreto presentan periodicidad de $2\pi$ en la frecuencia angular $\cos[(\omega\pm r 2\pi)k]=\cos[\omega k]$; con $r,k \in \mathbb{Z}$.

- A la expresión anterior se le conoce comunmente como la trasformada de Fouirer en tiempo discreto (Discrete time Fourier transform - DTFT).
y su respectiva transformada inversa (iDTFT) se expresa como:


\begin{equation}
x[k] = \frac{1}{2\pi}\int_{-\pi}^{+\pi}{X(\omega)e^{j\omega k}d\omega}=\frac{1}{2\pi}\int_{0}^{+2\pi}{X(\omega)e^{j\omega k}d\omega}
\end{equation}


# Transformada discreta de Fourier y transformada rápida de Fourier

- Sin embargo, tanto para la serie de Fourier, como para la transformada (continua y discreta en tiempo), se require de la resolución analítica de integrales; además, de la suposición de intervalos infinitos en tiempo y frecuencia (para el caso de la transformada de Fourier), los cuales no son viables ni implementables en la práctica.

- Por consiguiente, el concepto de espectro en frecuencia se extiende a sistemas de proceso de digital (tiempo discreto y  frecuencia discreta), mediante el cálculo de la transformada discreta de Fourier (Discrete Fourier Transform - DFT).

- En este caso, la transformada discreta se define como:

\begin{equation}
X[n] = \sum_{k=0}^{K-1}{x[k]e^{-jk\frac{2\pi n}{K}}},
\end{equation}

**Nota**: observe que el ángulo en la función euler $\theta_n = \frac{k 2\pi n}{K}$se restringe al intervalo $[0,2\pi)$ para $k\in\{0,1,\dots,K-1\}$, dado que se trabaja con funciones cos/sen en tiempo discreto.

- La DFT es una discretización en frecuencia de la DTFT:

\begin{equation}
X[n] = X(\omega)\Big|_{\omega=\frac{2\pi n}{K}},
\end{equation}

- Respecto a la serie de Fourier, la DFT se puede relacionar mediante:

\begin{equation}
X[n] = K c_n,
\end{equation}
asumiendo señal en tiempo discreto y $\omega_o = \frac{2\pi}{K}$ en representación por serie.

- Por otra parte, la transforma discreta de Fourier inversa (iDFT) se define como:

\begin{equation}
x[k] = \sum_{n=0}^{K-1}{X[n]e^{jk\frac{2\pi n}{K}}}.
\end{equation}

Nuevamente, note que las variables tiempo y frecuencia discretas $n,k\in\{0,1,\dots,K-1\}.$ Dado que la frecuencia fundamental se restringe como $\omega_o=\frac{2\pi}{K},$ las bases de representación en tiempo discreto codificaran frecuencias angulares en el rango $[0,2\pi)$ (cosenos y senos originales).

- No obstante, en la práctica el cálculo de las sumatorias es costoso computacionalmente, a medida que $K$ aumenta. Por consiguiente, se cuentan con paquetes computacionales que optimizan el cálculo de la DFT de forma rápida, lo que conlleva a la comunmente conocida transformada rápida de Fourier ([Fast Fourier Transform - FFT](https://es.wikipedia.org/wiki/Transformada_r%C3%A1pida_de_Fourier)).

# Ejemplo:

- Se presenta el cálculo del espectro mediante DFT (`fft` de numpy) para la señal:


\begin{equation}
x(t) = \begin{cases} A, & -T/2 \leq t < 0 \\ -A, & 0 \leq t < T/2 \end{cases}
\end{equation}

```python
# Para graficar
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import warnings
warnings.filterwarnings("ignore")

mpl.rc('axes', labelsize=14)
mpl.rc('xtick', labelsize=12)
mpl.rc('ytick', labelsize=12)

A = 1 # amplitud de la señal
T = 5 # segundos de gráfica
Fs = 1000 # frecuencia de muestreo definida
tv = np.arange(-T/2,T/2,1/Fs) #vector de tiempo generado
Nm = len(tv) #numero de muestras requeridas

#se crea la señal A y -A concatenando
xo = np.concatenate((A*np.ones(int(Nm/2)), -A*np.ones(int(Nm/2))), axis=0)
#graficar señal original
plt.plot(tv,xo,'r',linewidth=4)
plt.grid()
plt.xlabel("t[s]",fontsize = 14)
plt.ylabel("x(t)",fontsize = 14)
plt.show()
```

```python
#cálculo de la FFT de la señal signo invertido
vf = np.fft.fftfreq(len(xo),1/Fs) #calculo vector de frecuencias
Xw = np.fft.fft(xo) #cálculo FFT

plt.stem(vf,abs(Xw),'r')
plt.title(r'|X[n]|')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|$',fontsize = 14)
plt.axis('tight')
plt.grid()
plt.show()
plt.stem(vf,abs(Xw)/len(vf),'r')
plt.title(r'|X[n]|/K=c_n')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|/K$',fontsize = 14)
plt.show()
```

**Nota**: La función `np.fft.fftfreq` cálcula el equivalente en Hz de cada armónico, teniendo en cuenta que en tiempo discreto:

$$\theta_n = \frac{k 2\pi n}{K} = k \omega_n = k n \omega_o$$

Además, en el proceso de discretización ($F_s$ frecuencia de muestreo):

$$\omega_n t = 2\pi F_n t =  \frac{2\pi F_n k}{F_s} = 2 \pi f n k = \frac{2 \pi n k}{K}$$

donde $f=\frac{1}{K}$ es equivalente a la frecuencia en Hz en tiempo discreto.

Así:

$$\frac{2\pi F_n k}{F_s}=\frac{2 \pi n k}{K}$$

y:

$$F_n = \frac{F_s n}{K} [Hz]$$




- Note que:

$x(t) = \frac{1}{2\pi}\int_{-\infty}^\infty X(\omega) e^{j\omega t} d\omega$

$x^*(t) = \left( \frac{1}{2\pi}\int_{-\infty}^\infty X(\omega) e^{j\omega t} d\omega\right)^*=\frac{1}{2\pi}\int_{-\infty}^\infty X^*(\omega) e^{-j\omega t} d\omega$

Haciendo $\omega' = - \omega$:


$x^*(t) = \frac{1}{2\pi}\int_{-\infty}^\infty X^*(-\omega') e^{j\omega' t} d\omega'$

Entonces:

$\mathscr{F}\{x^*(t)\} = X^*(-\omega) = \mathscr{Re}\{X(-\omega) \} -j\mathscr{Im}\{X(-\omega) \} $

Y

$\mathscr{F}\{x(t)\} = X(\omega) = \mathscr{Re}\{X(\omega) \} +j\mathscr{Im}\{X(\omega) \} $

Por lo tanto:

$\mathscr{Re}\{X(\omega) \}=\mathscr{Re}\{X(-\omega) \}$
$\mathscr{Im}\{X(\omega) \}=-\mathscr{Im}\{X(-\omega) \}$

-Dado que el espectro es simétrico para señales reales ($x(t)\in \mathbb{R}$), se sugiere utilizar la expresión simplificada `rfft`:

```python
vf = np.fft.rfftfreq(len(xo),1/Fs) #calculo vector de frecuencias
Xw = np.fft.rfft(xo) #cálculo FFT
plt.stem(vf,abs(Xw),'r')
plt.title(r'|X[n]|')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|$',fontsize = 14)
plt.axis('tight')
plt.grid()
plt.show()
plt.stem(vf,abs(Xw)/(2*len(vf)),'r')
plt.title(r'|X[n]|/K=c_n')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|/K$',fontsize = 14)
plt.grid()
plt.show()
```

- La reconstrucción de x[k] se puede obtener mediante la transformada inversa:

```python
xe = np.fft.irfft(Xw) #transformada inversa
plt.plot(tv,xe)
plt.title(r'Reconstrucción completa')
plt.xlabel(r't',fontsize = 14)
plt.ylabel(r'$x[n]$',fontsize = 14)
plt.show()
```

- Se pueden escoger o ponderar los coeficientes para filtrar (alterar) la reconstrucción de la señal.

```python
Xwf = Xw.copy()
Xwf[10:] = 0 #solo se tienen en cuenta los 10 primeros armónicos
xe2 = np.fft.irfft(Xwf)
plt.plot(tv,xe2)
plt.show()
```

  - Filtro pasa-bajas (low-pass)
  - Filtro pasa-altas (high-pass)
  - Filtro pasa-banda (band-pass)
  - Filtro rechaza banda (band-stop)

Presente y describa el modelo que utiliza la FFT para el cálculo de los armónicos.

Descripción del modelo:

1. Señal de Entrada:
- Se parte de una señal de entrada que puede ser una función temporal (por ejemplo, una grabación de audio) representada en el dominio del tiempo. Esta señal puede contener una mezcla de múltiples frecuencias.

2. Aplicación de la FFT:
- La FFT se aplica a la señal de entrada. Este algoritmo eficiente convierte la señal del dominio del tiempo al dominio de la frecuencia. En términos más técnicos, la FFT calcula los coeficientes de la serie de Fourier discreta (DFT), que corresponden a las amplitudes y fases de las ondas sinusoidales que componen la señal original.

3. Identificación de los Armónicos:
- En el dominio de la frecuencia, los armónicos son las componentes que aparecen en múltiplos de la frecuencia fundamental (la frecuencia más baja presente en la señal). La frecuencia fundamental se encuentra como el pico de mayor energía en el espectro de frecuencia.
- Los armónicos son entonces los picos sucesivos en múltiplos de esta frecuencia fundamental.

4. Análisis de los Armónicos:
- Se pueden analizar las amplitudes y fases de los armónicos para comprender las características de la señal original, como su timbre en el caso de señales musicales o las distorsiones en el caso de señales eléctricas.


Aplicaciones Comunes:

- Análisis de Señales Eléctricas: En el análisis de sistemas eléctricos, los armónicos se analizan para identificar y mitigar distorsiones en la señal eléctrica.
- Análisis de Audio: En la música, los armónicos determinan el timbre de los instrumentos, y su análisis es fundamental para la síntesis de sonidos y el procesamiento de señales de audio.


Ventajas del Modelo:

- Eficiencia Computacional: La FFT es una versión optimizada de la DFT, lo que permite realizar cálculos rápidos incluso para señales de gran tamaño.
- Descomposición Completa: Permite una descomposición completa de la señal en sus componentes de frecuencia, facilitando la identificación y análisis de los armónicos.


Limitaciones:

- Resolución de Frecuencia: La resolución de frecuencia depende del tamaño de la ventana de tiempo analizada. Señales con componentes muy cercanas en frecuencia pueden no ser distinguibles si la resolución es baja.
- Suposición de Periodicidad: La FFT supone que la señal es periódica dentro de la ventana de análisis, lo que puede introducir artefactos en señales no periódicas.




#SOLUCION PUNTO 2


| **Aspecto**                | **DFT**                                            | **FFT**                                              |
|---------------------------|----------------------------------------------------|------------------------------------------------------|
| **Complejidad**           | $$O(N^2)$$. | $$O(N \log N)$$                                    |
| **Descripción**           | Calculada mediante sumatoria directa, ineficiente para señales grandes. | Optimiza el cálculo con "divide y vencerás", eficiente para señales grandes. |
| **Aplicación**            | Adecuada para señales de longitud pequeña a moderada. | Preferida en procesamiento de señales y análisis de audio. |

Con esta comprensión teórica de las diferencias en complejidad y eficiencia entre la DFT por sumatoria y la FFT, podemos proceder a la simulación práctica para observar estas diferencias en acción.

```python
import numpy as np
import time
import matplotlib.pyplot as plt

# Función para calcular la DFT usando la definición directa (sumatoria)
def dft_sum(signal):
    N = len(signal)
    n = np.arange(N)
    k = n.reshape((N, 1))
    M = np.exp(-2j * np.pi * k * n / N)
    return np.dot(M, signal)

# Función para calcular la FFT usando la función de numpy
def fft(signal):
    return np.fft.fft(signal)

# Valores de longitud de la señal de entrada a probar
lengths = [4, 8, 16, 32, 64, 128, 256, 512, 1024, 2048, 4096]

# Arrays para almacenar los tiempos de cómputo
dft_times = []
fft_times = []

# Cálculo de los tiempos de cómputo para cada longitud de señal
for length in lengths:
    signal = np.random.random(length)

    # Tiempo para la DFT
    start_time = time.time()
    dft_sum(signal)
    dft_times.append(time.time() - start_time)

    # Tiempo para la FFT
    start_time = time.time()
    fft(signal)
    fft_times.append(time.time() - start_time)

# Graficar los resultados
plt.figure(figsize=(10, 6))
plt.plot(lengths, dft_times, label='DFT (Sumatoria)', marker='o')
plt.plot(lengths, fft_times, label='FFT (Numpy)', marker='o')
plt.yscale('log')
plt.xscale('log')
plt.xlabel('Longitud de la señal de entrada')
plt.ylabel('Tiempo de ejecución (segundos)')
plt.title('Comparación del tiempo de cómputo: DFT vs FFT')
plt.legend()
plt.grid(True)
plt.show()
```

#SOLUCION PUNTO 3

1. ¿Qué significa -3 dB?
El decibelio (dB) es una unidad logarítmica que se usa para expresar relaciones de potencia o amplitud. En el contexto de filtros, se suele referir a la magnitud de la función de transferencia o respuesta en frecuencia del sistema.

- Una atenuación de -3 dB significa que la potencia de la señal ha disminuido al 50 % de su valor máximo.

- En términos de amplitud (por ejemplo, voltaje), esto corresponde a una caída a aproximadamente el 70.7 % del valor máximo, ya que:

$$20 \log_{10}(0.707) \approx -3\,\text{dB}$$

2. Frecuencia de -3 dB
- Es la frecuencia en la cual la respuesta en magnitud de un sistema o filtro cae 3 dB por debajo del valor máximo (típicamente el valor en la banda pasante).

- Se utiliza como un punto de referencia para indicar el límite entre la banda pasante y la banda atenuada del filtro.

3. Relación con las frecuencias de corte
- En un filtro (por ejemplo, pasa bajos, pasa altos, pasa banda), la o las frecuencias de corte se definen como aquellas donde la magnitud cae a -3 dB del valor máximo.

- Estas frecuencias determinan los límites del espectro de frecuencia que el filtro permite pasar con mínima atenuación.

```python
import yt_dlp
import subprocess
import os
import numpy as np
import soundfile as sf
import matplotlib.pyplot as plt
from scipy.fft import fft, ifft, fftfreq

# Paso 1: Descargar audio y recortar a 5 segundos en WAV 44.1 kHz mono
def download_audio_5s(video_url, output_name="clip5s.wav"):
    output_file = "temp_audio.%(ext)s"
    ydl_opts = {
        'format': 'bestaudio/best',
        'outtmpl': output_file,
        'postprocessors': [],  # No postprocessing, keep original format
        'prefer_ffmpeg': True,
        'overwrites': True
    }

    try:
        print(f"🔻 Descargando audio desde: {video_url}")
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            info = ydl.extract_info(video_url, download=True)
            actual_file = ydl.prepare_filename(info)
        print(f"✅ Descarga completada: {actual_file}")

        print(f"🎧 Procesando audio con ffmpeg (recorte a 5s y conversión a WAV mono 44.1kHz)...")
        subprocess.run(
            ["ffmpeg", "-y", "-i", actual_file, "-t", "30", "-ac", "1", "-ar", "44100", output_name],
            check=True,
            capture_output=True,
            text=True
        )
        print("✅ Conversión exitosa.")
        os.remove(actual_file)

    except subprocess.CalledProcessError as e:
        print(f"❌ Error en ffmpeg:\n{e.stderr}")
        raise e
    except Exception as e:
        print(f"❌ Error inesperado: {e}")
        raise e

# Paso 2: Aplicar filtros FFT
def apply_filter_fft(data, samplerate, filter_type, f_low=None, f_high=None):
    N = len(data)
    fft_data = fft(data)
    freqs = fftfreq(N, d=1/samplerate)
    mask = np.ones(N, dtype=bool)

    if filter_type == "lowpass":
        if f_high is None:
            raise ValueError("f_high must be specified for lowpass")
        mask[np.abs(freqs) > f_high] = False
    elif filter_type == "highpass":
        if f_low is None:
            raise ValueError("f_low must be specified for highpass")
        mask[np.abs(freqs) < f_low] = False
    elif filter_type == "bandpass":
        if f_low is None or f_high is None:
            raise ValueError("f_low and f_high must be specified for bandpass")
        mask[(np.abs(freqs) < f_low) | (np.abs(freqs) > f_high)] = False
    elif filter_type == "bandstop":
        if f_low is None or f_high is None:
            raise ValueError("f_low and f_high must be specified for bandstop")
        mask[(np.abs(freqs) > f_low) & (np.abs(freqs) < f_high)] = False
    else:
        raise ValueError(f"Filtro desconocido: {filter_type}")

    filtered_fft = fft_data * mask
    filtered_signal = np.real(ifft(filtered_fft))
    return filtered_signal

# Paso 3: Ejecutar todo
video_url = "https://www.youtube.com/watch?v=CHXzLUzzW5c"

try:
    # Descargar y recortar audio
    download_audio_5s(video_url)

    # Leer señal
    data, samplerate = sf.read("clip5s.wav")
    print(f"📥 Archivo leído: 'clip5s.wav' ({samplerate} Hz, {data.shape})")

    # Si es estéreo, usar primer canal
    if data.ndim > 1:
        print("🎚️ Audio estéreo. Usando canal izquierdo.")
        data_mono = data[:, 0]
    else:
        data_mono = data

    # Aplicar filtros
    print("🔧 Aplicando filtros FFT...")
    low = apply_filter_fft(data_mono, samplerate, "lowpass", f_high=1000)
    high = apply_filter_fft(data_mono, samplerate, "highpass", f_low=2000)
    band = apply_filter_fft(data_mono, samplerate, "bandpass", f_low=500, f_high=3000)
    stop = apply_filter_fft(data_mono, samplerate, "bandstop", f_low=1000, f_high=2000)
    print("✅ Filtros aplicados.")

    # Guardar audios filtrados
    sf.write("lowpass.wav", low, samplerate)
    sf.write("highpass.wav", high, samplerate)
    sf.write("bandpass.wav", band, samplerate)
    sf.write("bandstop.wav", stop, samplerate)
    print("💾 Archivos de audio filtrado guardados.")

    # Graficar resultados
    print("📊 Generando gráficas...")
    time_vector = np.arange(len(data_mono)) / samplerate

    plt.figure(figsize=(12, 8))

    plt.subplot(2,2,1)
    plt.title("Filtro Pasa Bajas")
    plt.plot(time_vector, low)
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Amplitud")

    plt.subplot(2,2,2)
    plt.title("Filtro Pasa Altas")
    plt.plot(time_vector, high)
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Amplitud")

    plt.subplot(2,2,3)
    plt.title("Filtro Pasa Bandas")
    plt.plot(time_vector, band)
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Amplitud")

    plt.subplot(2,2,4)
    plt.title("Filtro Rechaza Banda")
    plt.plot(time_vector, stop)
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Amplitud")

    plt.tight_layout()
    plt.show()
    print("✅ Proceso completo.")

except Exception as e:
    print(f"💥 Falló el proceso completo: {e}")
```
