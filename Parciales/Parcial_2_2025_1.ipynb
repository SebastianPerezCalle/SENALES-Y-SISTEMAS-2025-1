Pregunta 1 – Sistema Masa–Resorte–Amortiguador
1. Modelo físico

  El sistema mostrado en la Figura 1 está compuesto por:

  - Masa: \( m \)
  - Amortiguador: \( b \)
  - Resorte: \( k \)
  - Entrada: fuerza externa:    F_E(t)
  - Salida: posición de la masa:   y(t)


2. Ecuación dinámica del sistema.

  Aplicando la segunda ley de Newton al cuerpo:

    $$
    F_E(t) = m \ddot{y}(t) + b \dot{y}(t) + k y(t)
    $$


3. Transformada de Laplace

  Bajo condiciones iniciales cero:

    $$
    F_E(s) = m s^2 Y(s) + b s Y(s) + k Y(s)
    $$

  Factorizando \( Y(s) \):

    $$
    F_E(s) = \left( m s^2 + b s + k \right) Y(s)
    $$

4. Función de transferencia en lazo abierto

    $$
    G(s) = \frac{Y(s)}{F_E(s)} = \frac{1}{m s^2 + b s + k}
    $$

  La función de transferencia del sistema masa–resorte–amortiguador es:

    $$
    G(s) = \frac{1}{m s^2 + b s + k}
    $$

  - La entrada del sistema es la fuerza aplicada \( F_E(t) \)
  - la salida es el desplazamiento \( y(t) \).
  - Función de tranferencia en lazo abierto:

**Sistema Eléctrico Equivalente (Figura 2)**

1. Analogía electro-mecánica

    Se establece una analogía entre el sistema masa–resorte–amortiguador y un circuito RLC serie, como se muestra a continuación:

    

| Sistema Mecánico             | Sistema Eléctrico (RLC serie)       |
|-----------------------------|-------------------------------------|
| Fuerza \( F(t) \)           | Voltaje \( V_{in}(t) \)             |
| Masa \( m \)                | Inductancia \( L \)                 |
| Amortiguador \( b \)        | Resistencia \( R \)                 |
| Resorte \( k \)             | Inverso de la capacitancia \( \frac{1}{C} \) |
| Velocidad \( \dot{y}(t) \)  | Corriente \( i(t) \)                |
| Posición \( y(t) \)         | Carga \( v_0(t) \)                    |

    

2. Ecuación del circuito RLC serie

  Aplicando la Ley de Kirchhoff de Voltajes para una malla RLC serie:

    $$
    V_{in}(t) = v_R(t) + v_L(t) + v_C(t)
    $$

  Donde:

    $$
    v_R(t) = R i(t), \quad
    v_L(t) = L \frac{di(t)}{dt}, \quad
    v_C(t) = \frac{1}{C} \int i(t)\, dt
    $$

  Derivando ambos lados para eliminar la integral:

    $$
    \frac{dV_{in}(t)}{dt} = R \frac{di(t)}{dt} + L \frac{d^2 i(t)}{dt^2} + \frac{1}{C} i(t)
    $$

  Reorganizando:

    $$
    L \frac{d^2 i(t)}{dt^2} + R \frac{di(t)}{dt} + \frac{1}{C} i(t) = \frac{dV_{in}(t)}{dt}
    $$

3. Transformada de Laplace

  Asumiendo condiciones iniciales nulas:

    $$
    L s^2 I(s) + R s I(s) + \frac{1}{C} I(s) = s V_{in}(s)
    $$

  Factorizando:

    $$
    I(s) \left( L s^2 + R s + \frac{1}{C} \right) = s V_{in}(s)
    $$

4. Función de transferencia del circuito  
    Si la salida es la corriente \( i(t) \):
    
    $$
    \frac{I(s)}{V_{in}(s)} = \frac{s}{L s^2 + R s + \frac{1}{C}}
    $$

    Si la salida es el voltaje del condensador \( V_C(s) \), que es lo más común para hacer la equivalencia con la posición:
    
    $$
    G(s) = \frac{V_C(s)}{V_{in}(s)} = \frac{1}{L C s^2 + R C s + 1}
    $$


  - La función de transferencia del sistema eléctrico equivalente al sistema masa–resorte–amortiguador es:  

  
  $$
    G(s)  = \frac{1}{L C s^2 + R C s + 1}
    $$
    

  - Y su correspondencia directa con el sistema mecánico es:

$$
\boxed{
\frac{1}{m s^2 + b s + k} \quad \longleftrightarrow \quad \frac{1}{L C s^2 + R C s + 1}
}
$$

  - Esta analogía permite modelar un sistema mecánico como un circuito eléctrico equivalente, lo cual facilita su análisis con herramientas de electrónica o control.

```python
# Panel Interactivo – Simulación de Sistemas Dinámicos en Streamlit

# Este código crea a interactive dashboard that allows the user to analyze
# different configurations of second-order systems (masa-spring-damper or RLC circuit),
# both in open-loop and closed-loop. It allows visualizing temporal responses,
# Bode diagrams, poles and ceros, and estimating physical parameters of the system.

%%writefile Punto1.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as signal # Import scipy.signal for numerical control functions

# Configuración del layout del panel
st.set_page_config(layout="wide")
st.title(" Panel Interactivo – Análisis de Sistemas Dinámicos")

# Selección del tipo de sistema (condición de amortiguamiento)
tipo = st.selectbox("Selecciona el tipo de respuesta del sistema:", [
    "Subamortiguada", "Sobreamortiguada", "Críticamente amortiguada", "Inestable"])

# Selección de factor de amortiguamiento según el tipo seleccionado
if tipo == "Subamortiguada":
    zeta = st.slider("Factor de amortiguamiento (ζ)", 0.01, 0.99, 0.5)
elif tipo == "Críticamente amortiguada":
    zeta = 1.0
elif tipo == "Sobreamortiguada":
    zeta = st.slider("Factor de amortiguamiento (ζ)", 1.01, 5.0, 1.5)
else:
    zeta = st.slider("Factor de amortiguamiento (ζ)", -2.0, -0.01, -0.5)

# Frecuencia natural del sistema
wn = st.slider("Frecuencia natural ωₙ [rad/s]", 0.1, 20.0, 5.0)

# Define the transfer function coefficients
num = [1]
den = [1, 2*zeta*wn, wn**2]

# Create transfer function object using scipy.signal
system = signal.TransferFunction(num, den)

# Closed-loop system (assuming unity feedback)
# Manual calculation of closed-loop transfer function coefficients
num_closed = num
den_closed = [den[0], den[1] + num[0], den[2]] # For unity feedback: G_closed = G_open / (1 + G_open)

system_closed = signal.TransferFunction(num_closed, den_closed)


# Estimación de componentes físicos (masa-resorte-amortiguador y RLC)
m = 1                                             # Masa asumida igual a 1 kg
k = wn**2 * m                                     # Constante del resorte
b = 2 * zeta * wn * m                             # Coeficiente de amortiguamiento
L = m                                             # Inductancia equivalente
R = b                                             # Resistencia equivalente
C = 1 / k                                         # Capacitancia equivalente

# Visualización de componentes estimados
st.sidebar.subheader(" Componentes estimados")
st.sidebar.markdown(f"**Masa m:** {m:.2f} kg")
st.sidebar.markdown(f"**Amortiguador b:** {b:.2f} Ns/m")
st.sidebar.markdown(f"**Resorte k:** {k:.2f} N/m")
st.sidebar.markdown(f"**R:** {R:.2f} Ω")
st.sidebar.markdown(f"**L:** {L:.2f} H")
st.sidebar.markdown(f"**C:** {C:.4f} F")

# Tiempo de simulación
t = np.linspace(0, 10, 1000)

# Respuesta al impulso, escalón y rampa in closed loop using scipy.signal
t_imp, y_imp = signal.impulse(system_closed, T=t)
t_step, y_step = signal.step(system_closed, T=t)

# Generate ramp input
t_ramp = np.linspace(0, 10, 1000)
ramp_input = t_ramp

# Forced response to ramp using scipy.signal.lsim
t_lsim, y_ramp, _ = signal.lsim(system_closed, U=ramp_input, T=t_ramp)


# Calculation of poles and zeros
poles_open = system.poles
zeros_open = system.zeros

# Calculation of frequency response for Bode plot
w_bode, mag, phase = signal.bode(system)


# Gráficos de análisis en lazo abierto
col1, col2 = st.columns(2)

with col1:
    st.subheader(" Diagrama de Bode (Lazo Abierto)")
    fig1, (ax_mag, ax_phase) = plt.subplots(2, 1)
    ax_mag.semilogx(w_bode, mag)
    ax_mag.set_ylabel('Magnitude [dB]')
    ax_mag.grid(True)
    ax_phase.semilogx(w_bode, phase)
    ax_phase.set_xlabel('Frequency [rad/s]')
    ax_phase.set_ylabel('Phase [deg]')
    ax_phase.grid(True)
    st.pyplot(fig1)

with col2:
    st.subheader(" Polos y ceros (Lazo Abierto)")
    fig2, ax2 = plt.subplots()
    ax2.plot(np.real(poles_open), np.imag(poles_open), 'x', label='Poles')
    ax2.plot(np.real(zeros_open), np.imag(zeros_open), 'o', label='Zeros')
    ax2.set_xlabel('Real')
    ax2.set_ylabel('Imaginary')
    ax2.set_title('Pole-Zero Map')
    ax2.grid(True)
    ax2.axhline(0, color='black', linewidth=0.5)
    ax2.axvline(0, color='black', linewidth=0.5)
    ax2.legend()
    st.pyplot(fig2)

# Gráficos de respuestas temporales en lazo cerrado
st.subheader(" Respuestas del sistema (Lazo Cerrado)")
col3, col4, col5 = st.columns(3)

with col3:
    st.markdown("**Respuesta al impulso**")
    fig3, ax3 = plt.subplots()
    ax3.plot(t_imp, y_imp)
    ax3.set_xlabel("Tiempo [s]")
    ax3.set_ylabel("Amplitud")
    ax3.grid()
    st.pyplot(fig3)

with col4:
    st.markdown("**Respuesta al escalón**")
    fig4, ax4 = plt.subplots()
    ax4.plot(t_step, y_step)
    ax4.set_xlabel("Tiempo [s]")
    ax4.set_ylabel("Amplitud")
    ax4.grid()
    st.pyplot(fig4)

with col5:
    st.markdown("**Respuesta a la rampa**")
    fig5, ax5 = plt.subplots()
    ax5.plot(t_lsim, y_ramp)
    ax5.set_xlabel("Tiempo [s]")
    ax5.set_ylabel("Amplitud")
    ax5.grid()
    st.pyplot(fig5)

# Visualización de parámetros temporales de desempeño
# scipy.signal does not have a direct equivalent of ctl.step_info that returns a dictionary
# We can calculate some metrics manually if needed, but for now, skip this part or provide basic info
st.subheader("📋 Parámetros temporales (Escalón)")
# You can add manual calculation of metrics here based on y_step and t_step
# For example:
# peak_value = np.max(y_step)
# st.write(f"Peak Value: {peak_value:.2f}")
st.write("Step response metrics can be calculated manually from the step response data.")

# The ScriptRunContext warning can be ignored when running in bare mode like in Colab
```

**Modulación y Demodulación SSB-AM**

La modulación por amplitud en banda lateral única (SSB-AM) es una técnica eficiente que transmite solo una de las bandas laterales (superior o inferior) del espectro de una señal AM, eliminando la portadora y la otra banda lateral para ahorrar ancho de banda y potencia.


- **Modular**  
  Significa variar una propiedad de una señal postadora usando una señal de información.  

  Ejemplo:
  En modulación AM (amplitud modulada), se varía la amplitud de una señal de alta frecuencia (portadora) según el valor de la señal de mensaje.  

  - Señal de mensaje: m(t) (audio, voz, datos).
  - Señal portadora: $$c(t) = A_c \cos(2\pi f_c t)$$

  Modulación AM:
    $$s(t) = [A_c+m(t)] \cos(2\pi f_c t)$$
    En SSB, FM, PM y otras modulaciones, se modifica una propiedad (amplitud, frecuencia o fase).
- **Desmodular**

  Consiste en recuperar la señal original a partir de una señal modulada que fue transmitida.

**Modela Matemático:**  
  - **Señal moduladora**  
  Sea la señal de mensaje: $$m(t)$$  

  - **Portadora**  
  Una señal senoidal de alta frecuencia: $$c(t) = A_c \cos(2\pi f_c t)$$  

  - **Señal AM convencional (DSB-AM)**
  $$s_AM(t) = [A_c+m(t)]\cos(2\pi f_c t)$$  

  - **Señal DSB-SC (doble banda lateral suprimida portadora):**  
  $$s_{\text{DSB-SC}}(t) = m(t) \cos(2\pi f_c t)$$  

  - **Señal SSB-AM (Banda lateral Única):**  
  Para la banda lateral superior (USB)
  $$s_{\text{DSB}}(t) = m(t) \cos(2\pi f_c t) - \hat{m}(t) \sin(2\pi f_c t)$$  
  Para la banda lateral inferior (LSB)  
  $$s_{\text{DSB}}(t) = m(t) \cos(2\pi f_c t) + \hat{m}(t) \sin(2\pi f_c t)$$  

  Donde:  
  $$\hat{m}(t)$$ es la transformada de Hilbert de $$m(t)$$

  - Esta operación genera un desfase de 90° que permite suprimir una de las bandas laterales.  


**Modelo en el dominio de la frecuencia (Transformada de fourier)**  
  - **Transformada de la señal baseband**  
  $$M(f) = \mathcal{F}\{m(t)\}$$  
  Espectro centrado en f = 0, de ancho B.  

  - **Transformada de SSB(USB):**

  $$S_{\text{SSB}}(f) =
  \begin{cases}
  M(f - f_c), & f > f_c \quad \text{(USB)} \\
  0, & \text{en otro caso}
  \end{cases}$$  

  - **Transformada de SSB (LSB):**
  $$S_{\text{SSB}}(f) =
  \begin{cases}
  M(f + f_c), & f > -f_c \quad \text{(USB)} \\
  0, & \text{en otro caso}
  \end{cases}$$  

**Demodulación**  

  Se multiplica la señal recibida por la portadora original:  
  $$r(t) = s_{\text{SSB}}(t) \cdot \cos(2\pi f_c t)$$  

Y luego se pasa por un filtro pasa-bajas (LPF) para recuperar
𝑚(𝑡).

### Modulación en el dominio del tiempo y frecuencia

| **Concepto**           | **Expresión en el tiempo**                                                              | **Transformada de Fourier**                                             |
|------------------------|------------------------------------------------------------------------------------------|-------------------------------------------------------------------------|
| **Señal baseband**     | \( m(t) \)                                                                               | \( M(f) \)                                                              |
| **DSB-SC**             | \( m(t)\cos(2\pi f_c t) \)                                                               | \( \frac{1}{2} \left[ M(f - f_c) + M(f + f_c) \right] \)                |
| **SSB (USB)**          | \( m(t)\cos(2\pi f_c t) - \hat{m}(t)\sin(2\pi f_c t) \)                                  | \( M(f - f_c), \quad f > 0 \)                                           |
| **SSB (LSB)**          | \( m(t)\cos(2\pi f_c t) + \hat{m}(t)\sin(2\pi f_c t) \)                                  | \( M(f + f_c), \quad f < 0 \)                                           |
| **Demodulación coherente** | Multiplicar por \( \cos(2\pi f_c t) \) y filtrar                                      | Centrar espectro y aplicar filtro pasa-bajas                           |

```python
# ssb_am_dashboard.py

%%writefile Punto2.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, lfilter, freqz, tf2zpk
from scipy.fftpack import fft, fftfreq
from scipy.io import wavfile
import soundfile as sf
import io

st.set_page_config(layout="wide")


# FUNCIONES AUXILIARES


def butter_bandpass(lowcut, highcut, fs, order=5):
    nyq = 0.5 * fs  # Frecuencia de Nyquist
    low = lowcut / nyq
    high = highcut / nyq
    b, a = butter(order, [low, high], btype='band')
    return b, a

def apply_filter(data, b, a):
    return lfilter(b, a, data)

def plot_freq_domain(signal, fs, title="Espectro"):
    N = len(signal)
    f = fftfreq(N, 1/fs)
    S = np.abs(fft(signal)) / N
    plt.figure()
    plt.plot(np.fft.fftshift(f), np.fft.fftshift(S))
    plt.title(title)
    plt.xlabel('Frecuencia [Hz]')
    plt.ylabel('Magnitud')
    plt.grid()
    st.pyplot(plt.gcf())
    plt.close()

def plot_time(signal, t, title="Señal en el tiempo"):
    plt.figure()
    plt.plot(t, signal)
    plt.title(title)
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Amplitud")
    plt.grid()
    st.pyplot(plt.gcf())
    plt.close()

def plot_bode(b, a, fs):
    w, h = freqz(b, a, worN=8000)
    freqs = 0.5 * fs * w / np.pi
    plt.figure()
    plt.semilogx(freqs, 20 * np.log10(abs(h)))
    plt.title('Diagrama de Bode')
    plt.xlabel('Frecuencia [Hz]')
    plt.ylabel('Magnitud [dB]')
    plt.grid()
    st.pyplot(plt.gcf())
    plt.close()

def plot_pzmap(b, a):
    z, p, _ = tf2zpk(b, a)
    plt.figure()
    plt.scatter(np.real(z), np.imag(z), marker='o', label='Ceros')
    plt.scatter(np.real(p), np.imag(p), marker='x', label='Polos')
    plt.axhline(0, color='black')
    plt.axvline(0, color='black')
    plt.title('Plano de polos y ceros')
    plt.xlabel('Re')
    plt.ylabel('Im')
    plt.grid()
    plt.legend()
    st.pyplot(plt.gcf())
    plt.close()


# INTERFAZ PRINCIPAL


st.title("Dashboard Interactivo - Modulación y Demodulación SSB-AM")

st.markdown("""
Este panel permite visualizar el proceso completo de modulación y demodulación de una señal en **Banda Lateral Única (SSB-AM)**.
Puedes elegir entre dos tipos de señal de mensaje:
- Pulso rectangular.
- Segmento de audio (archivo de 5 segundos).

También puedes ver cómo operan los filtros digitales IIR usados en el proceso (diagrama de Bode y plano de polos y ceros).
""")


# PARÁMETROS INICIALES


fs = 44100  # Frecuencia de muestreo
fc = 10000  # Frecuencia de portadora

tipo_senal = st.selectbox("Selecciona tipo de señal mensaje:", ["Pulso rectangular", "Audio (5 segundos)"])


# GENERAR SEÑAL DE MENSAJE


if tipo_senal == "Pulso rectangular":
    duracion = 0.05  # segundos
    t = np.linspace(0, duracion, int(fs * duracion), endpoint=False)
    mensaje = np.ones_like(t)
    mensaje[int(0.3 * len(t)):] = 0  # Rectángulo
    st.markdown("### Señal de mensaje (pulso rectangular)")
    plot_time(mensaje, t, "Mensaje - Tiempo")
    plot_freq_domain(mensaje, fs, "Mensaje - Frecuencia")

elif tipo_senal == "Audio (5 segundos)":
    audio_file = st.file_uploader("Carga un archivo de audio WAV de 5 segundos:", type=["wav"])
    if audio_file is not None:
        data, fs = sf.read(audio_file)
        if len(data.shape) > 1:
            data = data[:, 0]  # usar solo canal izquierdo
        mensaje = data[:fs * 5]
        t = np.linspace(0, len(mensaje) / fs, len(mensaje), endpoint=False)
        st.markdown("### Señal de mensaje (audio)")
        plot_time(mensaje, t, "Mensaje - Tiempo")
        plot_freq_domain(mensaje, fs, "Mensaje - Frecuencia")
    else:
        st.stop()


# MODULACIÓN SSB


from scipy.signal import hilbert

# Transformada de Hilbert para generar señal en cuadratura
m_hat = np.imag(hilbert(mensaje))
portadora_cos = np.cos(2 * np.pi * fc * t)
portadora_sin = np.sin(2 * np.pi * fc * t)

ssb_usb = mensaje * portadora_cos - m_hat * portadora_sin  # USB
st.markdown("### Señal modulada SSB (USB)")
plot_time(ssb_usb, t, "SSB-AM - Tiempo")
plot_freq_domain(ssb_usb, fs, "SSB-AM - Frecuencia")


# DEMODULACIÓN


producto = ssb_usb * 2 * np.cos(2 * np.pi * fc * t)

# Filtro pasa bajos (para recuperar señal)
b, a = butter(6, 4000 / (0.5 * fs), btype='low')
recuperada = lfilter(b, a, producto)

st.markdown("### Señal demodulada")
plot_time(recuperada, t, "Señal demodulada - Tiempo")
plot_freq_domain(recuperada, fs, "Señal demodulada - Frecuencia")


# ANÁLISIS DEL FILTRO IIR


st.markdown("### Análisis del filtro IIR usado")
plot_bode(b, a, fs)
plot_pzmap(b, a)

st.markdown("""
**Notas:**
- Se ha usado la transformada de Hilbert para generar la componente en cuadratura necesaria en la modulación SSB.
- La demodulación se realiza mediante detección coherente (multiplicación por portadora y filtrado).
- El filtro IIR es de orden 6, tipo Butterworth, pasa bajos.
""")
```
