```python
import soundfile as sf # para instalar pip install soundfile
import numpy as np # Se agrega la importación de numpy
import os # Se agrega la importación de os

#leer archivos y crear np.array audios
path = 'results/'
wav_files = [f for f in os.listdir(path) if f.endswith('.wav')] # Asegurarse de que wav_files está definido si esta celda se ejecuta independientemente

# Leer el primer archivo para obtener la frecuencia de muestreo de referencia
if len(wav_files) > 0:
    # Se lee el primer archivo para obtener la frecuencia de muestreo
    x_first, fs_ref = sf.read(path + wav_files[0])
    # Usar fs_ref para calcular la dimensión del array x_t
    ts = 5 #t segmento
    samples_per_segment = int(ts * fs_ref)
    tl = np.array([40,50,60,70,80,90,100]) #puntos lectura
    Ns = len(wav_files) * len(tl) #cantidad segmentos
    x_t = np.zeros((Ns, samples_per_segment, x_first.shape[-1])) # Ns segmentos, cantidad de muestras, número de canales del primer archivo
else:
    print("No se encontraron archivos .wav en la carpeta 'results'.")
    # Inicializar x_t con dimensiones dummy o salir si no hay archivos
    samples_per_segment = 0
    Ns = 0
    x_t = np.zeros((0, 0, 0))


label = np.zeros((Ns,1)) #vector tipo de genero
name_c = []
#leer archivos wav
i = 0
for name in wav_files:#lectura audio .wav
    # Leer el archivo y su frecuencia de muestreo
    x, fs_current = sf.read(path + name)

    # Asegurarse de que la frecuencia de muestreo es la de referencia o manejar diferencias
    # Para este error particular, la causa es que la longitud del segmento extraído
    # es diferente a la esperada por x_t.
    # Asumiremos que todos los archivos tienen la misma fs_ref o cerca,
    # y que la diferencia causó el error.
    # Si las fs varían significativamente, se requeriría remuestreo.
    # Aquí, nos aseguramos que el número de muestras extraídas coincida con samples_per_segment

    for ti in tl: #segmentos de tiempo
        start_sample = int(fs_current * ti)
        end_sample = int(fs_current * (ti + ts))

        # Extraer el segmento
        segment = x[start_sample:end_sample, :]

        # Verificar la longitud del segmento extraído
        # Si la longitud no coincide con samples_per_segment (basado en fs_ref),
        # entonces la causa del error es la diferencia entre fs_current y fs_ref
        # o problemas con la duración exacta de los segmentos en los archivos.
        # Para solucionar el ValueError directamente, debemos asegurarnos
        # que 'segment' tenga la misma forma que x_t[i].
        # Esto implica que la longitud de 'segment' debe ser 'samples_per_segment'.

        # Si el segmento extraído es más corto (por ejemplo, al final del archivo)
        # o más largo debido a diferencias en fs, esto causará el error.
        # Una solución simple es pad/trim el segmento extraído a la longitud esperada.

        current_segment_length = segment.shape[0]
        if current_segment_length != samples_per_segment:
            # print(f"Warning: Segment length mismatch for {name} at time {ti}-{ti+ts}. Expected {samples_per_segment}, got {current_segment_length}.")
            # Si el segmento es más corto, rellenar con ceros
            if current_segment_length < samples_per_segment:
                padding = np.zeros((samples_per_segment - current_segment_length, x_first.shape[-1]))
                segment = np.vstack((segment, padding))
            # Si el segmento es más largo, recortar
            elif current_segment_length > samples_per_segment:
                segment = segment[:samples_per_segment, :]

        # Ahora 'segment' tiene la forma correcta (samples_per_segment, número de canales)
        x_t[i] = segment # Asigna el segmento ajustado a x_t
        # Asegurarse que el índice -5 sea correcto para extraer el tipo de género
        # Esto depende del formato del nombre del archivo 'banda_numero_tipo.wav'
        # Si el nombre es 'Zaider_14_2.wav', name[-5] es '2'. int('2') es 2.
        # Si el nombre es 'AnotherBand_1_1.wav', name[-5] es '1'. int('1') es 1.
        # Si el nombre es 'YetAnother_10_1.wav', name[-5] es '_'. Esto causaría un ValueError.
        # Si el tipo de género es un número de un solo dígito justo antes de '.wav'
        # y el nombre tiene un formato consistente, name[-5] debería ser el dígito.
        # Si no, se necesita una forma más robusta de extraer el número del tipo.
        # Por ahora, mantenemos name[-5] basado en el contexto original.

        try:
            label[i] = int(name[-5]) #tipo de genero
        except ValueError:
             print(f"Warning: Could not extract genre type from filename '{name}'. Check filename format.")
             # Asignar un valor por defecto o manejar el error según necesidad
             label[i] = -1 # Ejemplo: usar -1 para indicar un error o tipo desconocido


        name_c += [name[:-6]] # Extrae el nombre antes del '_' y el tipo numérico
        print(f"{i} lectura: {name}; segundo {ti}:{ti+ts}; tipo música {label[i]}")
        i+=1

# fs ahora contendrá la fs del último archivo leído.
# Si quieres la fs_ref usada para inicializar x_t, debes usar fs_ref.
# Para propósitos de FFT y visualización, es mejor usar la fs consistente
# usada para definir el número de muestras por segmento.
fs = fs_ref if 'fs_ref' in locals() else 48000 # Usar la frecuencia de referencia o un valor por defecto

print(x_t.shape)
```

```python
#calculo de fourier
vf = np.fft.rfftfreq(x_t.shape[1],1/fs) #calculo vector de frecuencias
Xw = np.fft.rfft(x_t,axis=1).mean(axis=-1) #transformada rapida de Fourier para señal Real a lo largo del tiempo (axis=1) y se promedian los dos canales
Xw.shape
```

```python
#grafica tiempo y fourier
plt.plot(np.arange(0,ts,1/fs),x_t.mean(axis=-1).T) #se promedian los dos canales stereo
plt.xlabel('t[s]')
plt.ylabel('x(t)')
plt.show()
```

```python
plt.plot(vf,abs(Xw).T)
plt.xlabel('f[Hz]')
plt.ylabel('|X(f)|')
plt.show()
```

```python
#se normalizan espectros entre 0 y 1 para evitar inconsistencias por ampliltudes máximas
from sklearn.preprocessing import MinMaxScaler
sca = MinMaxScaler()
Xw_ = sca.fit_transform(abs(Xw).T).T

plt.plot(vf,Xw_.T)
plt.xlabel('f[Hz]')
plt.ylabel('|X(f)|')
plt.show()

#en dB
plt.plot(vf,(20*np.log10(Xw_+1e-10)).T) # se suma 1e-10 para evitar discontinuidad del log
plt.xlabel('f[Hz]')
plt.ylabel('|X(f)| dB')
plt.show()
```
